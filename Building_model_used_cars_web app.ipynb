```python
x=df2.drop('Price',axis=1)
y=df2['Price']
```

```python
num_cols=x.select_dtypes(exclude='object').columns.tolist()
cat_cols=x.select_dtypes(include='object').columns.tolist()
print(num_cols)
print(cat_cols)
```

```python
nominal_cols=['Location','Fuel_Type','Transmission','Owner_Type','Brand','Model']
ordinal_cols=['Seats','Owner_Type']
```

```python
# pip install scikit-learn
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2,random_state=42)
x_train.shape, x_test.shape
```

```python
from sklearn.preprocessing import StandardScaler,OneHotEncoder,OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

num_pipline=Pipeline([
('num_imputer',SimpleImputer(strategy='median')),
('Scaler',StandardScaler())
])

nominal_pipline=Pipeline([
('num_imputer',SimpleImputer(strategy='most_frequent')),
('ohe',OneHotEncoder(drop='first',sparse_output=False,handle_unknown='ignore'))
])

ordinal_pipline=Pipeline([
('ord_imputer',SimpleImputer(strategy='most_frequent')),
('ohe',OrdinalEncoder())
])

preprocessor=ColumnTransformer([
            ('num_pipline',num_pipline,num_cols),
            ('nominal_pipline',nominal_pipline,nominal_cols),
            ('ordinal_pipline',ordinal_pipline,ordinal_cols),
            ]).set_output(transform='pandas')

preprocessor
```

```python
x_train_prep=preprocessor.fit_transform(x_train)
x_test_prep=preprocessor.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train_prep,y_train)
```

```python
model.intercept_
```

```python
model.coef_
```

```python
#Error and evaluation metrics
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
y_pred=model.predict(x_test_prep)
print('r2_score', r2_score(y_test,y_pred))
print('mean_squared_error',mean_squared_error(y_test,y_pred))
print('mean_absolute_error',mean_absolute_error(y_test,y_pred))
```

```python
# prameters are predicated. Is there a way to optimize the mode.
# Let's use Ridge to optimize the intercept
from sklearn.linear_model import Ridge
model=Ridge(alpha=0.1)
model.fit(x_train_prep,y_train)
y_pred=model.predict(x_test_prep)
print('r2_score',r2_score(y_test,y_pred))
print('mean_squared_error',mean_squared_error(y_test,y_pred))
print('mean_absolute_error',np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
# let's use try different degree of polynomial features and different alpha values for Ridge regression to find the best model using GridSearchCV

from sklearn.model_selection import GridSearchCV 
model=pipline = Pipeline([
    ('preprocessor', preprocessor),
    ('ridge', Ridge())
])  

param_grid = {
    'preprocessor__num_pipeline__poly__degree': [2,3,4,5],
    'ridge__alpha': [0.01, 0.1, 1]
}   

grid_search = GridSearchCV(model, param_grid, cv = 5, scoring = 'r2')
grid_search.fit(x_train, y_train)
```

```python
pd.DataFrame(grid_search.cv_results_)
```

```python
# Define the best model based on the grid search results, best ridge regression model with the best degree of polynomial features
grid_search.best_params_
```

```python
grid_search.best_score_
```

```python
#extract the best model (parameters) from the grid search
best_model = grid_search.best_estimator_
best_model.predict(x_test)
print('r2_score', r2_score(y_test, best_model.predict(x_test)))
```

```python
# extract and save the best model using joblib. The file will be saved in the current working directory as 'linear_regression_model.pkl'
# This is the output of the model model file which can be used for deployment in production environment. The model can be loaded using joblib.load('linear_regression_model.pkl') and used for predictions on new data.
import joblib
best_model = grid_search.best_estimator_
joblib.dump(best_model, 'linear_regression_model.pkl')
print('Linear Regression Model saved successfully!')
```
